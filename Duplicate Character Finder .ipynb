```python
text = input("Enter a word or sentence: ").lower()

# Dictionary to count frequencies of each letter
char_counts = {}

for char in text:
    if char.isalpha():  # Only process letters, ignore spaces and punctuation
        char_counts[char] = char_counts.get(char, 0) + 1

# Filter for characters that appear more than once
duplicates = {char: count for char, count in char_counts.items() if count > 1}

print("\nDuplicate Letters:")
if duplicates:
    for char, count in duplicates.items():
        print(f"  '{char}' appears {count} times")
else:
    print("  No duplicate letters found!")
```

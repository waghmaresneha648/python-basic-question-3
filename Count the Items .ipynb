```python
def count_items(arr):
    # 1. Initialize tracker to 0
    total_count = 0
    
    for item in arr:
        # 2. Add 1 to tracker for each item seen
        total_count += 1
        
    return total_count

# Test Case
inventory = ["shoes", "shirt", "hat", "socks", "jacket"]
print("Total items:", count_items(inventory))  # Should print: 5
```
